```python
from pathlib import Path
import json
import pandas as pd

import matplotlib.pyplot as plt

total_logs_path = Path("total_logs.json")
new_log_path = Path("log.json")

# Read existing total_logs.json if it exists
if total_logs_path.exists():
    try:
        total_logs_df = pd.read_json(total_logs_path)
    except ValueError:
        total_logs_df = pd.read_json(total_logs_path, lines=True)
else:
    total_logs_df = pd.DataFrame()

# Get the insertId values already in total_logs.json
existing_ids = set(total_logs_df["insertId"].astype(str)) if not total_logs_df.empty else set()

# Read existing log.json if it exists
if new_log_path.exists():
    try:
        log_df = pd.read_json(new_log_path)
    except ValueError:
        log_df = pd.read_json(new_log_path, lines=True)
else:
    log_df = pd.DataFrame()

# Filter log.json to keep only records with insertId not in total_logs.json
new_records = log_df[~log_df["insertId"].astype(str).isin(existing_ids)]

# Append new records to total_logs.json
if not new_records.empty:
    total_logs_df = pd.concat([total_logs_df, new_records], ignore_index=True)

# Save to total_logs.json in lines format
total_logs_df.to_json(total_logs_path, orient="records", lines=True)

print(f"Added {len(new_records)} new records to total_logs.json")
print(f"Total records in total_logs.json: {len(total_logs_df)}")

insert_raw = total_logs_df["insertId"].astype(str)

# Try direct datetime parse first
insert_dt = pd.to_datetime(insert_raw, errors="coerce")

# Fallback 1: date pattern YYYY-MM-DD
mask = insert_dt.isna()
insert_dt.loc[mask] = pd.to_datetime(
    insert_raw.loc[mask].str.extract(r"(\d{4}-\d{2}-\d{2})", expand=False),
    errors="coerce"
)

# Fallback 2: compact date pattern YYYYMMDD
mask = insert_dt.isna()
insert_dt.loc[mask] = pd.to_datetime(
    insert_raw.loc[mask].str.extract(r"(\d{8})", expand=False),
    format="%Y%m%d",
    errors="coerce"
)

hits_per_day_insertid = (
    total_logs_df.assign(insert_date=insert_dt)
    .dropna(subset=["insert_date"])
    .assign(day=lambda d: d["insert_date"].dt.date)
    .groupby("day")
    .size()
    .sort_index()
)

plt.figure(figsize=(12, 6))
plt.plot(hits_per_day_insertid.index, hits_per_day_insertid.values, marker="o")
plt.title("Number of Hits per Day (from insertId)")
plt.xlabel("Day")
plt.ylabel("Number of Hits")
plt.xticks(rotation=45, ha="right")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

```
